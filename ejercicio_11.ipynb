#### a) ¿Por qué la reconstrucción tomográfica se considera un problema inverso? Identificar qué información se conoce, qué variable se busca recuperar y qué papel cumple el modelo $b = Ax$.

Se considera un problema inverso porque lo que se conoce no es la causa, sino sus efectos. Es decir, el escáner no mide la imagen directamentesino cuánto se atenúa cada rayo al atravesar el objeto. A partir de esas mediciones hay que deducir cómo es el objeto por dentro, es decir deducir los efectos a la causa.

- **Se conoce de antemano:** las mediciones $b$, una por cada rayo, que juntas forman el sinograma, y la matriz $A$, que describe qué píxeles atraviesa cada rayo y en qué medida.
- **Lo que se busca deducir:** la imagen $x$, es decir, el coeficiente de atenuación de cada píxel.

Por otro lado el modelo $b = Ax$ es lo que conecta las mediciones con la imagen. Sale de la ley de atenuación de la consigna, $I = I_0\, e^{-\int \mu}$. A

Al tomar logaritmo, cada medición pasa a ser la suma de las atenuaciones de los píxeles que atraviesa el rayo. Por ende, cada fila del sistema $b = Ax$ es una ecuación que dice "la suma de estos píxeles da esta medición".

El problema directo es calcular las mediciones $b$ a partir de una imagen $x$ conocida, ( un producto matriz-vector), como se hizo en los ejercicios 2 y 3. El problema inverso es recuperar $x$ a partir de $b$ lo cual es mucho más difícil porque requiere analizar si la información alcanza, qué hacer cuando no hay solución exacta y cómo evitar que el ruido arruine el resultado. Eso es lo que se estudió a lo largo de todo el trabajo.

#### b) ¿Cómo puede determinarse, a partir de $A$, si las mediciones permiten recuperar una única imagen? Relacionar la respuesta con el rango y el espacio nulo de $A$, y considerar qué ocurre cuando existe $v \neq 0$ tal que $Av = 0$.

Para saber si una imagen se puede replicar de forma unica hay que mirar el rango de $A$. Como $A$ tiene una columna por píxel ($n^2$ columnas, 64 en el fantoma), por el teorema del rango y nulidad

$$\text{rango}(A) + \dim(\text{Nul}(A)) = n^2$$

Si $x$ es una solución de $Ax = b$, todas las soluciones son de la forma $x + \text{Nul}(A)$, es decir, una solución particular más las soluciones del sistema homogéneo $Av = 0$. Por ende, la imagen es única solo si el espacio nulo contiene únicamente al vector cero

$$\text{solución única} \iff \text{Nul}(A) = \{0\} \iff \text{rango}(A) = n^2$$

Las columnas de $A$ tienen que ser linealmente independientes. No alcanza con tener muchas filas, porque algunas mediciones pueden ser redundantes. 

Si existe $v \neq 0$ con $Av = 0$, entonces $\text{rango}(A) < n^2$ y, por linealidad, para cualquier $\gamma$

$$A(x + \gamma v) = Ax + \gamma\,Av = b + 0 = b$$

Es decir, hay infinitas imágenes distintas que producen exactamente las mismas mediciones, y a partir de $b$ no hay forma de saber cuál es la real. Como $a_r^T v = 0$ para cada fila, $v$ es un patrón que suma cero a lo largo de todos los rayos, por lo que el escáner no lo puede ver. Esto se vio en el ejercicio 4, al sumarle al fantoma un vector del espacio nulo de $A_4$ se obtuvo una imagen claramente distinta con las mismas 46 mediciones ($\|A_4 x_{\text{alt}} - b_4\| \approx 10^{-15}$). También se vio en el Ejercicio 5, donde el residuo fue prácticamente cero en los tres casos, pero el error relativo fue 0.47 con $A_2$, 0.14 con $A_4$ y 0 con $A_7$. Por lo tanto, solo cuando el espacio nulo es $\{0\}$ un residuo nulo garantiza haber recuperado la imagen real.

#### c) ¿Cómo afecta el ruido al sistema de ecuaciones? Explicar qué problema resuelve el método de mínimos cuadrados, interpretar el residuo obtenido y discutir si su norma es suficiente para evaluar la calidad de una reconstrucción.

Sin ruido, $b = Ax$ pertenece a $\text{Col}(A)$ y el sistema tiene solución exacta. Con ruido, $b_\sigma = Ax + \eta$ en general queda fuera de $\text{Col}(A)$, y el sistema $Ax = b_\sigma$ se vuelve inconsistente, ninguna imagen reproduce exactamente las mediciones, tal como se explico en el ejercicio 6.

Como no hay solución exacta, se busca la imagen cuyas mediciones están más cerca de las medidas utilizando minimos cuadrados

$$\hat{x} = \arg\min_x \|Ax - b_\sigma\|^2$$

 $A\hat{x}$ es la proyección ortogonal de $b_\sigma$ sobre $\text{Col}(A)$. En el ejercicio 7 se verificó que $\|A_7^T r_\sigma\| \approx 10^{-14}$, es decir, que el residuo es ortogonal a todas las columnas de $A_7$, que es la misma condición que dan las ecuaciones normales $A^TA\hat{x} = A^Tb_\sigma$.

El residuo $r = b_\sigma - A\hat{x}$ es la parte de las mediciones que ninguna imagen puede explicar. Como mide si la imagen reproduce las mediciones y no si se parece a la imagen real, no es suficiente para evaluar la calidad de una reconstruccion. A lo largo del trabajo se pudo ver de la siguiente forma:

- en el Ejercicio 5, las reconstrucciones con $A_2$ y $A_4$ tuvieron residuo ≈ 0 y aun así errores grandes, por el espacio nulo
- en el Ejercicio 9, al aumentar $\lambda$ el residuo siempre creció pero con ruido el error bajó
- en el Ejercicio 10, un residuo mucho menor que la norma del ruido mostro que la reconstrucción se estaba ajustando al ruido.


#### d) ¿Qué supuesto sobre la imagen introduce el término $\|Dx\|^2$? Analizar cómo cambian las reconstrucciones al modificar $\lambda$ y justificar las diferencias observadas.

 El término $\|Dx\|^2$ supone que la imagen real es "suave", es decir que los píxeles vecinos tienen valores parecidos y que los cambios bruscos solo ocurren en los bordes.

$\lambda$ indica muestra el trade off entre suavidad de la medicion y la fidelidad

$$J_\lambda(x) = \frac{1}{2}\left(\|Ax - b\|^2 + \lambda\,\|Dx\|^2\right)$$

- Con $\lambda$ chico: la reconstrucción se ajusta a las mediciones, incluido el ruido, y se ve granulada (en el ejercicio 10, con $\lambda = 0$, aparecieron valores de hasta −0.54).
- Con $\lambda$ mediano: se reduce el ruido sin perder la forma de la imagen. En el ejercicio 9 el error bajó de 0.163 a 0.138 con $\sigma = 0.10$, y de 0.325 a 0.231 con $\sigma = 0.20$.
- Con $\lambda$ demasiado grande: la suavidad pesa más que las mediciones y se pierden los contornos reales, la imagen queda borrosa y el error vuelve a subir (con $\lambda = 1$ en el Ejercicio 9, y con $\lambda = 100$ en el 10).



#### e) Comparar `np.linalg.lstsq`, gradiente descendiente y Newton considerando las iteraciones necesarias, los parámetros que requiere cada método y las soluciones obtenidas. Explicar el comportamiento observado para Newton.

Los tres métodos resuelven el mismo problema, minimizar $E(x) = \frac{1}{2}\|Ax - b\|^2$, pero de formas distintas. `lstsq` es un método directo, es decir, no itera, calcula la solución de una sola vez y no hay que elegir ningún parámetro. Gradiente descendiente, en cambio, se acerca al mínimo de a poco usando solo el gradiente, por lo que suele necesitar muchas iteraciones. Es el único que depende de parámetros que hay que elegir, la tasa de aprendizaje $\alpha$,el limite de convergencia y un máximo de iteraciones. La elección de $\alpha$ es clave, ya que el método converge solo si $0 < \alpha < 2/\mu_{\max}$, donde $\mu_{\max}$ es el mayor autovalor de la Hessiana. Si $\alpha$ es muy chico avanza lento y si es muy grande se puede pasar del intervalo y diverger. Por último, Newton usa el gradiente y la Hessiana y no tiene parámetros que elegir, pero necesita que $H = A^TA$ sea invertible, es decir, que $A$ tenga unicamente columnas linealmente independientes.Cuando $A$ tiene rango completo, los tres llegan a la misma solución, porque minimizan la misma función y esa función tiene un solucion unica. 

Newton llega en una sola iteración porque aproxima la función con su desarrollo de Taylor de segundo orden y salta directamente al mínimo de esa aproximación. Como $E(x)$ ya es una función cuadrática con Hessiana constante no devuelve una aproximaciónsino la función. Por lo que el mínimo que encuentra es el mínimo real. De hecho, desde cualquier punto inicial $x_0$

$$x_1 = x_0 - (A^TA)^{-1}A^T(Ax_0 - b) = x_0 - x_0 + (A^TA)^{-1}A^Tb = (A^TA)^{-1}A^Tb$$

que es justamente la solución de las ecuaciones normales.

Sin embargo, que Newton y `lstsq` resuelvan el problema de una vez no significa que siempre convengan. Al Newton necesitar la Hessiana $H = A^TA$, que contiene una fila y una columna por cada píxel, y en cada paso resolver un sistema con ella. Por ejemplo para una imagen de $128 \times 128$ eso representa una matriz de $16384 \times 16384$, es decir más de 268 millones de entradas que hay que armar, guardar en memoria y usar para resolver el sistema. Algo parecido ocurre con `lstsq`, que necesita $A$ como matriz completa, por lo que no aprovecha que la mayoría de sus entradas son cero. Esto es justamente lo que la consigna del Ejercicio 10 pide evitar, ya que $A$ no debe convertirse en una matriz densa.

Gradiente descendiente, en cambio, nunca necesita armar $H$. En cada iteración solo calcula el gradiente $A^T(Ax - b)$, que son dos productos matriz-vector. Primero $Ax$ y después $A^T$ por el resultado. Como $A$ es dispersa, cada rayo atraviesa pocos píxeles, por lo que cada fila tiene pocos valores distintos de cero, y esos productos solo usan esos valores. Por ende,aunque haga falta hacer muchas iteraciones cada una de ellas es barata. Por eso es el método que se usa en imágenes grandes, como en el ejercicio 10, donde incluso $\mu_{\max}$ se calculó con el método de potencias, que también usa solo productos con $A$ y $A^T$ sin armar la Hessiana.



#### f) Elaborar una síntesis que conecte el modelo $b = Ax$ con el rango, el espacio nulo, las proyecciones ortogonales, mínimos cuadrados y la regularización. Indicar qué aspecto del problema de reconstrucción permite analizar cada concepto.

Todo el trabajo parte del modelo $b = Ax$, que describe la tomografía como una transformación lineal, es decir, cada medición es la suma de las atenuaciones de los píxeles que atraviesa un rayo. Para reconstruir la imagen se conocen las mediciones $b$ y la matriz $A$, y se busca poder reconstruir la imagen $x$.

El **rango** y el **espacio nulo** permiten analizar si las mediciones alcanzan para determinar una única imagen(Suponiendo que las mediciones no contienen ruido). Por el teorema del rango y nulidad, la imagen es única solo si el espacio nulo contiene únicamente al vector cero, es decir si el rango es igual a la cantidad de píxeles. Si existe $v \neq 0$ con $Av = 0$, $v$ es un patrón que el escáner no puede ver, y hay infinitas imágenes compatibles con las mismas mediciones. Por eso la cantidad y la dirección de los rayos importan más que la cantidad de mediciones.

**Mínimos cuadrados** permite analizar qué hacer cuando las mediciones tienen ruido. En ese caso $b$ queda fuera de $\text{Col}(A)$ y el sistema no tiene solución exacta por lo que se busca la imagen que minimiza $\|Ax - b\|^2$. Esa imagen cumple las ecuaciones normales $A^TAx = A^Tb$, y se puede calcular de forma directa o con métodos iterativos como gradiente descendiente y Newton, cuyo comportamiento depende de los autovalores de $A^TA$.

Las **proyecciones ortogonales** permiten interpretar qué significa esa solución. $A\hat{x}$ es la proyección ortogonal de $b$ sobre $\text{Col}(A)$, es decir las mediciones más parecidas a $b$ que sí puede producir alguna imagen. El residuo es lo que queda afuera, ortogonal a todas las columnas de $A$, y representa la parte de las mediciones que ninguna imagen puede explicar.

Por último, la **regularización** permite analizar el trade off entre fidelidad a las mediciones, reducción del ruido y preservación de detalles. Mínimos cuadrados no distingue la información real del ruido y termina copiándolo. Al agregar $\lambda\|Dx\|^2$ se supone que la imagen es suave, y $\lambda$ decide cuánto pesa esa suposición frente a las mediciones. Con un $\lambda$ muy chico la imagen conserva el ruido, y con uno muy grande se pueden perder informacion importante.
